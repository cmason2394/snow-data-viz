## combine all the files into one dataframe
Process each file so they have unique identifiers (adding the year column) and are in the correct format (melt so in long format). 
Takes the steps tried on one file above, and combines them into a function and loop to process all the files that way.

```python
# create lists of the files to import
snowfall_file_list = ['2015-snowfall.csv', 
                      '2016-snowfall.csv', 
                      '2017-snowfall.csv', 
                      '2018-snowfall.csv', 
                      '2019-snowfall.csv', 
                      '2020-snowfall.csv', 
                      '2021-snowfall.csv', 
                      '2022-snowfall.csv',
                      '2023-snowfall.csv',
                      '2024-snowfall.csv',
                      '2025-snowfall.csv'
                     ]

snowdepth_file_list = ['2015-snowdepth.csv', 
                      '2016-snowdepth.csv', 
                      '2017-snowdepth.csv', 
                      '2018-snowdepth.csv', 
                      '2019-snowdepth.csv', 
                      '2020-snowdepth.csv', 
                      '2021-snowdepth.csv', 
                      '2022-snowdepth.csv',
                      '2023-snowdepth.csv',
                      '2024-snowdepth.csv',
                      '2025-snowdepth.csv'
                     ]

print(snowfall_file_list)
print(snowdepth_file_list)
```

```python
# save directory where files are stored
file_directory = 'data/'
```

```python
# create function to convert to dataframes, add year column, melt, and save as one dataframe
def process_snow_data(file_directory, file_list, measured_metric):
    all_df_list = []
    for filename in file_list:
        filepath = os.path.join(file_directory, filename)
        #print(f"Attempting to process: {filepath}")  # Debugging print
        try:
            year = filename.split('-')[0]
            df = pd.read_csv(filepath, skiprows=4)
            df['Year'] = int(year)
            melted_df = pd.melt(df,
                                 id_vars = ['Year', 'GHCN ID', 'Station Name', 'County', 'State', 'Elevation', 'Latitude', 'Longitude'],
                                 var_name = 'Date',
                                 value_name = measured_metric)
            all_df_list.append(melted_df)
        except FileNotFoundError:
            print(f"Error: File not found at {filepath}")
        except Exception as e:
            print(f"An error occurred while processing {filename}: {e}")
    combined_df = pd.concat(all_df_list, ignore_index=True)
    return combined_df
       
        
    
```

```python
# run function to create all-encompassing dataframes for snowfall and snowdepth
snowfall_df = process_snow_data(file_directory, snowfall_file_list, 'Snowfall (in)')
snowdepth_df = process_snow_data(file_directory, snowdepth_file_list, 'Snowdepth (in)')

print(snowfall_df.columns)
print(snowfall_df.head())
print(snowfall_df.info())
print(snowdepth_df.columns)
print(snowdepth_df.head())
print(snowdepth_df.info())
```

## Clean data

### Format date

```python
# function to format the date
def format_date(dataframe):
    # convert date column to string type
    dataframe['Date'] = dataframe['Date'].astype(str)

    # create full date string by reformatting as MonthAbbr-day-Year
    dataframe['Full_Date'] = dataframe['Date'].str.split(' ').str[1] + '-' + dataframe['Date'].str.split(' ').str[0] + '-' + dataframe['Year'].astype(str)

    # convert column to datetime object
    dataframe['Full_Date'] = pd.to_datetime(dataframe['Full_Date'], format='%d-%b-%Y')
```

```python
# run date formating function on dataframes
format_date(snowdepth_df)
format_date(snowfall_df)

print(snowdepth_df['Full_Date'])
print(snowfall_df['Full_Date'])
```

### handle missing values

```python
# see what types of values I have in the snow columns
print("Unique values in snowfall_df['Snowfall (in)']:")
print(snowfall_df['Snowfall (in)'].unique())

print("\nUnique values in snowdepth_df['Snow Depth (in)']:")
print(snowdepth_df['Snowdepth (in)'].unique())
```

```python
# replace 'M' with 'NaN'
snowfall_df['Snowfall (in)'].replace('M', pd.NA, inplace=True)
snowdepth_df['Snowdepth (in)'].replace('M', pd.NA, inplace=True)

print(snowfall_df['Snowfall (in)'].head(50))
print(snowdepth_df['Snowdepth (in)'].head(50))
```

```python
# handle trace values
snowfall_df['Snowfall (in)'].replace('T', 0, inplace=True)

print(snowfall_df['Snowfall (in)'].head(50))
```

```python
# convert columns to numeric datatype
snowfall_df['Snowfall (in)'] = pd.to_numeric(snowfall_df['Snowfall (in)'])
snowdepth_df['Snowdepth (in)'] = pd.to_numeric(snowdepth_df['Snowdepth (in)'])

print("Unique values in snowfall_df['Snowfall (in)'] after replacement:")
print(snowfall_df['Snowfall (in)'].unique())
print("\nData type of snowfall_df['Snowfall (in)']:")
print(snowfall_df['Snowfall (in)'].dtype)

print("Unique values in snowdepth_df['Snowdepth (in)'] after replacement:")
print(snowdepth_df['Snowdepth (in)'].unique())
print("\nData type of snowdepth_df['Snowdepth (in)']:")
print(snowdepth_df['Snowdepth (in)'].dtype)
```
